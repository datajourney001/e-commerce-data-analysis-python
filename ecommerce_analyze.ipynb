 ### Converting date columns 

```python
data['Order Date']=pd.to_datetime(data['Order Date'])
data['Ship Date']=pd.to_datetime(data['Ship Date'])
```

```python
# checking here
data.info()
```

```python
data.head()
```

```python
data['Order Month']=data['Order Date'].dt.month
data['order Year']=data['Order Date'].dt.year
data['Order Day Of Week']=data['Order Date'].dt.dayofweek
```

```python
data.head()
```

## EDA(Exploratory Data Analysis)

#### 1. Monthly Sales Analysis

```python
sales_by_month=data.groupby('Order Month')['Sales'].sum().reset_index()

sales_by_month
```

```python
#  seaborn bar chart

plt.figure(figsize=(10,6))
sns.barplot(x='Order Month',y='Sales',data=sales_by_month)
plt.title('Monthly Sales Analysis')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

### 2. Category-Wise Sales Analysis

```python
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()
sales_by_category
```

```python
# plotting pie chart using matplotlib

plt.figure(figsize=(8,5))
plt.pie(sales_by_category['Sales'],labels=sales_by_category['Category'],autopct='%1.1f%%')
plt.title('Category-wise Sales Analysis')
plt.show()
```

### 3. Sub-Category Sales Analysis

```python
sales_by_subcat = data.groupby('Sub-Category')['Sales'].sum().reset_index()
sales_by_subcat
```

```python
# barchart using seaborn

plt.figure(figsize=(8,5))
sns.barplot(x='Sales',y='Sub-Category',data=sales_by_subcat,hue='Sub-Category')
plt.title('Sub-Category-wise Sales Analysis')
plt.xlabel('Sub-Category')
plt.ylabel('Total Sales')
plt.show()
```

### 4.Monthly Profit Analysis

```python
profit_by_month=data.groupby('Order Month')['Profit'].sum().reset_index()
profit_by_month
```

```python
# linegraph using seaborn

sns.set_style('whitegrid')
plt.figure(figsize=(10,5))

# line graph
sns.lineplot(x='Order Month',y='Profit',data=profit_by_month)
plt.title('Monthly Profit Analysis')
plt.xlabel('Month')
plt.ylabel('Total Profit')
plt.show()
```

### 5. Analysis the profit by category and subcategory 

```python
profit_by_cat_sub=data.groupby(['Category','Sub-Category'])['Profit'].sum().reset_index()
profit_by_cat_sub
```

```python
plt.figure(figsize=(12,6))

# Profit by category and subcategory ( seaborn bar chart)

sns.barplot(x='Profit',y='Sub-Category',hue='Category',data=profit_by_cat_sub)
plt.title('Profit by Category & Subcategory')
plt.xlabel('Profit')
plt.ylabel('Sub-Category')
plt.show()
```

### 4. Customer Segment Analysis

```python
#sum of sales and profit on the basis of segment 

segment_data=data.groupby('Segment')[['Sales','Profit']].sum().reset_index()
segment_data
```

```python
segment_melted=segment_data.melt(id_vars='Segment',var_name='Metric',value_name='Amount')
segment_melted
```

```python
sns.set_theme(style='whitegrid')
plt.figure(figsize=(10,6))

sns.barplot(x='Segment',y='Amount',hue='Metric',data=segment_melted)
plt.title('Sales vs Profit by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Amount')
plt.show()
```

### 7. Sales to Profit-Ratio

```python
# prrofit margin Calculation 

total_sales=data['Sales'].sum()               # calculating >> total of  sales column 
total_profit=data['Profit'].sum()                    # calculating >> total of profit column
profit_margin= (total_profit/total_sales)*100

print(total_sales)
print(total_profit)
print(profit_margin)
```

```python
print("total sales",round(total_sales,2))
print("total profit",round(total_profit,2))
print("profit_margin",round(profit_margin,2))
```
